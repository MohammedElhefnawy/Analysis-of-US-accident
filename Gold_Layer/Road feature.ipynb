# The 13 road feature columns (Amenity, Bump, Crossing...) #

```python
import pandas as pd

df = pd.read_csv("Silver_Layer/US_Accidents_Silver.csv")

road_features = [
    'Amenity','Bump','Crossing','Give_Way','Junction',
    'No_Exit','Railway','Roundabout','Station','Stop',
    'Traffic_Calming','Traffic_Signal','Turning_Loop'
]

severity_map = {
    1: "Low Impact",
    2: "Moderate Impact",
    3: "High Impact",
    4: "Critical Impact"
}

rows = []

for feature in road_features:
    temp = df[df[feature] == 1].copy()

    grouped = (
        temp.groupby("Severity")
        .size()
        .reset_index(name="Accident_Count")
    )

    for _, r in grouped.iterrows():
        rows.append({
            "Road_Feature": feature,
            "Severity_Label": severity_map.get(r["Severity"], str(r["Severity"])),
            "Accident_Count": int(r["Accident_Count"])
        })

summary = pd.DataFrame(rows)

summary.to_csv(
    "Gold_Layer/Road_Feature_Severity.csv",
    index=False
)

print("Saved to Gold_Layer/Road_Feature_Severity.csv")
print(summary.to_string(index=False))
```
